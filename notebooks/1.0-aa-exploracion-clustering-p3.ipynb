# 1.0-aa-exploracion-clustering-p3

**Pregunta 3:** ¿Qué grupos de estaciones comparten patrones de uso y dónde se ubican en la ciudad?
**Objetivo:** Exploración de datos de viajes de MiBici y caracterización de estaciones para clustering (MLOps - Equipo 36).

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from mibici.config import PROCESSED_DATA_DIR, FIGURES_DIR
%matplotlib inline
sns.set_theme(style="whitegrid")
```

## 1. Carga de datos de viajes limpios y catálogo de estaciones

```
df_viajes = pd.read_parquet(PROCESSED_DATA_DIR / "viajes_limpios.parquet")
df_estaciones = pd.read_parquet(PROCESSED_DATA_DIR / "estaciones.parquet")
print(f"Total de viajes analizados: {len(df_viajes):,}")
print(f"Total de estaciones en catálogo: {len(df_estaciones):,}")
df_estaciones.head()
```

## 2. Inspección de Características de Perfilamiento por Estación (P3)

```
from mibici.features import calcular_perfil_estaciones_p3
df_features = calcular_perfil_estaciones_p3(df_viajes, df_estaciones)
df_features.describe()
```

## 3. Distribución de Variables Clave (Ratio de Flujo y Horas Pico)

```
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.histplot(df_features["ratio_flujo_neto"], kde=True, ax=axes[0], color="teal")
axes[0].set_title("Ratio de Flujo Neto ((Salidas - Llegadas)/Total)")

sns.histplot(df_features["pct_salidas_pico_matutino"], kde=True, ax=axes[1], color="orange")
axes[1].set_title("% Salidas en Pico Matutino (07-10h)")

sns.histplot(df_features["pct_salidas_pico_vespertino"], kde=True, ax=axes[2], color="purple")
axes[2].set_title("% Salidas en Pico Vespertino (17-20h)")
plt.tight_layout()
plt.show()
```
